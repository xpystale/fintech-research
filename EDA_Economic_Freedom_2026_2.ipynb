## 4. Distribution of the overall indicator of economic freedom

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    df["Overall Score"],
    bins=20,
    kde=True
)

plt.xlabel("Overall Score")
plt.ylabel("Number of countries")
plt.title("Overall Score Distribution")

plt.show()
```

The Overall Score is concentrated mainly between 50 and 70 points. Several countries have extremely low values, which creates a left tail in the distribution and suggests the presence of outliers.

## 6. Ranking of countries by level of economic freedom

```python
cols = ["Country", "Region", "Overall Score"]

display(df.nlargest(5, "Overall Score")[cols])
display(df.nsmallest(5, "Overall Score")[cols])
```

```python
extremes = pd.concat([
    df.nlargest(5, "Overall Score"),
    df.nsmallest(5, "Overall Score")
]).sort_values("Overall Score")

plt.figure(figsize=(10, 6))

sns.barplot(data=extremes, x="Overall Score", y="Country")

plt.xlim(0, 100)
plt.title("5 Highest and 5 Lowest EFE")
plt.show()
```

Singapore has the highest Overall Score (84.4), followed by Switzerland and Ireland. The lowest score is observed for North Korea (3.1), while Cuba and Venezuela also have substantially lower values than most countries in the dataset. The ranking demonstrates a large gap between countries with the highest and lowest levels of economic freedom.

## 7. Overall Score by Region

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="Overall Score",
    y="Region"
)

plt.title("Overall Score by Region")
plt.show()
```

```python
q1, q3 = df["Overall Score"].quantile([0.25, 0.75])
iqr = q3 - q1
outliers = (df["Overall Score"] < q1 - 1.5 * iqr) | (df["Overall Score"] > q3 + 1.5 * iqr)
```

```python
df.loc[outliers, ["Country", "Overall Score"]]
```

Outliers countries are explainable, no signs of error in the data have been identified.

## 8. Analysis of the components of economic freedom

```python
means = df[score_cols].mean().drop("Overall Score").sort_values()

plt.figure(figsize=(6, 6))
plt.scatter(means, means.index)
plt.axvline(df["Overall Score"].mean(), linestyle="--", label="Overall Score")

plt.title("Average Values of Economic Freedom Components")
plt.xlim(0, 100)
plt.legend(loc="lower right")
plt.show()
```

The economic freedom components show noticeable differences in their average values. Tax Burden has the highest mean score, while Government Integrity has the lowest.

## 9. Relationship between Financial Freedom and the Overall Score

```python
plt.figure(figsize=(10, 6))

sns.regplot(
    data=df,
    x="Financial Freedom",
    y="Overall Score",
)

plt.show()
```

```python
print(
    "Correlation:",
    df["Financial Freedom"].corr(df["Overall Score"])
)
```

Financial Freedom shows a strong positive relationship with the Overall Score. The correlation coefficient is approximately 0.81, indicating that countries with higher Financial Freedom scores tend to have higher overall levels of economic freedom. However, Financial Freedom is one of the components used to construct the Overall Score, so this relationship should be interpreted as an association within the index rather than as an independent causal effect.
